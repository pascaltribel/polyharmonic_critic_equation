```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from tqdm.auto import tqdm

torch.set_default_device('mps')
```

```python
eps = 1e-1
def objective(x, eps=eps):
    return x**2 + torch.sin(x/eps)**2
```

```python
plt.plot(torch.linspace(-5, 5, 1000).cpu(), objective(torch.linspace(-5, 5, 1000)).cpu())
```

```python
class Minimizer(nn.Module):
    def __init__(self):
        super().__init__()
        self.p = nn.Parameter(torch.ones((1,), requires_grad=True))

    def forward(self):
        return self.p
```

```python
n_epochs = 1000

m = Minimizer()
optimizer = optim.Adam(m.parameters(), lr=2*eps)
losses = []
for _ in (pbar:=tqdm(range(n_epochs))):
    for _ in range(10):
        optimizer.zero_grad()
        p = m()
        loss = objective(p)
        loss.backward()
        optimizer.step()
        losses.append(loss.item())
    pbar.set_description(f"Loss: {loss.item():.4e}, p: {p.item():.4e}")
```

```python
plt.plot(np.log(losses)[100:])
```
