```python
import numpy as np
from numpy.polynomial import legendre as L
from scipy.optimize import root
import matplotlib.pyplot as plt
from matplotlib import cm, colors

def solve(p=4.0, ell=3, N=40, Nq=400):
    x, w = L.leggauss(Nq)                       # quadrature nodes/weights
    V = L.legvander(x, N)                       # V[i,k] = P_k(x_i)
    k = np.arange(N + 1)
    lam = k * (k + 1.0)
    norm = (k + 0.5)                            # 1/||P_k||^2

    def F(c):
        u = V @ c
        f = np.abs(u)**(p - 2) * u
        return lam * c - norm * (V.T @ (w * f))

    def J(c):                                   # analytic Jacobian
        u = V @ c
        d = (p - 1) * np.abs(u)**(p - 2)
        return np.diag(lam) - (norm[:, None] * V.T) @ (w[:, None] * d[:, None] * V)

    # initial guess from projection onto P_ell
    Pl = V[:, ell]
    a = (ell*(ell+1) * np.sum(w*Pl**2) / np.sum(w*np.abs(Pl)**p))**(1/(p-2))
    c0 = np.zeros(N + 1); c0[ell] = a

    sol = root(F, c0, jac=J, method='lm', tol=1e-13)   # LM is robust if J is near-singular
    c = sol.x
    return c, sol, (x, V @ c)
```

```python
%matplotlib widget
def plot_sphere(U, TH, PH, title=""):
    """U, TH, PH: (n_theta, n_phi) arrays. Surface of the unit sphere colored by U."""
    Xs = np.sin(TH) * np.cos(PH)
    Ys = np.sin(TH) * np.sin(PH)
    Zs = np.cos(TH)

    vmax = np.max(np.abs(U))
    norm = colors.Normalize(vmin=-vmax, vmax=vmax)      # symmetric about 0
    m = cm.ScalarMappable(norm=norm, cmap="seismic")
    m.set_array(U)

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(projection="3d")
    ax.plot_surface(Xs, Ys, Zs, facecolors=m.to_rgba(U),
                    shade=False, rstride=1, cstride=1,
                    antialiased=False, linewidth=0)
    fig.colorbar(m, ax=ax, shrink=0.7, label="u")
    ax.set_box_aspect((1, 1, 1))
    ax.set_axis_off()
    ax.set_title(title)
    plt.tight_layout()
    plt.show()

def plot_axisymmetric(c, p=None, n=200):
    theta = np.linspace(0, np.pi, n)
    phi   = np.linspace(0, 2*np.pi, n)
    PH, TH = np.meshgrid(phi, theta)                    # shape (n_theta, n_phi)
    U = L.legval(np.cos(TH), c)                         # u = sum c_k P_k(cos theta)
    plot_sphere(U, TH, PH, title=f"p = {p}, ||u||_inf = {np.abs(U).max():.3g}")

# usage with the solver from before
c, sol, _ = solve(p=4, ell=3)
plot_axisymmetric(c, p=4)
```
